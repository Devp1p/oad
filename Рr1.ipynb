### Задание 3. Сортировка данных
Находим 10 треков с максимальным количеством прослушиваний (`stream_count`).

```python
top10 = df.sort_values('stream_count', ascending=False).head(10)
top10[['track_name', 'artist_name', 'genre', 'stream_count']]
```

**Ответ:** среди топ-10 встречаются жанры Alternative, Rock, Pop, Metal, Hip-Hop и R&B; исполнители: Grove Nova, Flora Quinn, Ethereal, SethStone, IsleEyre, Dean System, Rocco Ore, Knot Hart, The Inks и Prism Hart.

### Задание 4. Фильтрация
Находим треки, выпущенные после 2024 года.

```python
tracks_after_2024 = df[df['release_year'] > 2024]
tracks_after_2024
```

**Результат:** после 2024 года выпущено **8318 треков**; в датасете это записи 2025 года.

### Задание 4. Основные статистические показатели

```python
popularity_stats = df['popularity'].agg(['mean', 'median', 'min', 'max', 'std'])
popularity_stats
```

**Ответ:** среднее — **28.14**, медиана — **26**, минимум — **0**, максимум — **92**, стандартное отклонение — **16.00**.

### Задание 5. Мода

```python
print('Мода жанра:', df['genre'].mode().tolist())
print('Мода года выпуска:', df['release_year'].mode().tolist())
```

**Ответ:** наиболее часто встречающийся жанр — **R&B**, наиболее часто встречающийся год выпуска — **2024**.

### Задание 7. Гистограмма

```python
plt.figure(figsize=(9, 5))
plt.hist(df['popularity'], bins=10, range=(0, 100), edgecolor='black')
plt.title('Распределение популярности треков')
plt.xlabel('Популярность')
plt.ylabel('Количество треков')
plt.xlim(0, 100)
plt.show()
```

**Ответ по графику:** основная часть значений находится примерно в диапазоне **10–50**, особенно много значений в диапазоне **10–40**. Значения выше 70 встречаются значительно реже; значения 90–100 — крайне редко.
